# The Collatz Detour

A self-contained notebook: no data files, no network, no installs.

The Collatz conjecture says every positive integer eventually reaches 1 under
the rule *halve if even, triple-plus-one if odd*. Nobody has proved it. We are
not going to prove it here either — we are going to look at how long the
detour takes.

```python
def collatz_steps(n: int) -> int:
    """Number of steps for n to reach 1."""
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps
```

```python
print(f"27 takes {collatz_steps(27)} steps")
worst = max(range(1, 1000), key=collatz_steps)
print(f"worst under 1000: {worst} at {collatz_steps(worst)} steps")
```
